```python
from nilearn.datasets import MNI152_FILE_PATH

print(f"Path to template: {MNI152_FILE_PATH!r}")
```

```python
from nilearn import plotting

plotting.plot_img(MNI152_FILE_PATH)
```

### Image maniplulation

```python
# Smoothing
from nilearn import image
smoothed_img = image.smooth_img(MNI152_FILE_PATH, fwhm=8)
plotting.plot_img(smoothed_img)

another_smoothed_img = image.smooth_img(smoothed_img, fwhm=4)
plotting.plot_img(another_smoothed_img)
```

### Globbing over multiple 3D volumes
We can smooth volumes from many subjects at once, and get a 4D image as an output


```python
from nilearn import datasets
haxby = datasets.fetch_haxby(subjects=[1, 2])
```

```python
from pathlib import Path
anats_all_subjects = (
    Path(datasets.get_data_dirs()[0]) / "haxby2001" / "subj*"  / "anat*"
)
anats_all_subjects_smooth = image.smooth_img(anats_all_subjects, fwhm=5)
print(f"Smoothed anat images: {anats_all_subjects_smooth}")
```
